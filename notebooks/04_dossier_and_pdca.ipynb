## 4.1 PDCA iterations

```python
from biossbd import Cycle, Iteration, PDCA_CHECKLIST
pd.DataFrame(PDCA_CHECKLIST).T
```

```python
cycle = Cycle()
cycle.add(Iteration(label="v0 concept", phase="Plan",
                    kpis={"isi_normalised": 0.36, "hazardous_inventory_kg": 755, "ghg_intensity": 0.030}))
cycle.add(Iteration(label="v1 smaller holder", phase="Check",
                    kpis={"isi_normalised": 0.36, "hazardous_inventory_kg": 455, "ghg_intensity": 0.031},
                    changes=["Gas holder inventory halved"]))
cycle.check("v0 concept", "v1 smaller holder")
```

```python
print("Check gate passed:", cycle.gate("v0 concept", "v1 smaller holder"))
```
