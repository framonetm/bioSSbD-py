# 3. Evaluation: compare alternatives without hiding trade-offs

**Learning goals**
- Read results dimension by dimension (the default in the revised EC SSbD framework).
- Identify Pareto-optimal alternatives and explicit trade-offs.
- Use aggregation (MCDA) only as a sensitivity layer, and test robustness by Monte Carlo.

The numbers reproduce the published AcoD screening study exactly (see `tests/test_evaluation.py`).

```python
from pathlib import Path
import pandas as pd
import biossbd
from biossbd import load_system
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
EXAMPLE = Path("..") / "examples" / "acod" / "acod_system.yaml"
system = load_system(EXAMPLE)
print("bioSSbD-py", biossbd.__version__, "|", system.name)
```

```python
from biossbd import Evaluation
ev = Evaluation.from_system(system)
ev.dimension_scores((1, 5)).round(2)
```

```python
ax = ev.dimension_scores((1, 5)).T.plot.bar(figsize=(8, 4), rot=0, ylim=(0, 5))
ax.set_ylabel("score (1 worst - 5 best)"); ax.legend(fontsize=8)
```

## 3.1 Pareto front and trade-offs

```python
print("Pareto-optimal:", ev.pareto_front())
ev.tradeoffs()
```

```python
ev.criteria_summary()
```

## 3.2 Aggregation as a sensitivity check

```python
ev.compare_methods().round(3)
```

```python
print("Environmental weight at which the top alternative changes:", ev.weight_threshold("Environmental"))
```

## 3.3 Monte Carlo robustness
Ordinal scores move ±1 step (p = 0.25/0.5/0.25); weights are drawn from Dirichlet distributions.

```python
r = ev.robustness(n=5000, perturb_scores=True, perturb_weights=True)
r.rank_acceptability.round(3)
```

```python
r.order_probability.head().round(3)
```
