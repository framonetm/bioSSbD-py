# 1. Scoping analysis: defining the SSbD system

**Learning goals**
- Describe an innovation as an *SSbD system*: substances, processes (units), life-cycle stages and actors.
- Build the **life-cycle safety matrix** (stage × hazard × receptor).
- Find **scoping gaps** before any detailed assessment starts.

The revised EC SSbD framework (JRC, 2025) makes the Scoping Analysis the entry point of every assessment. Its three elements are (1) the SSbD system, (2) the intended innovation and (3) the actors along the life cycle. This notebook uses the anaerobic co-digestion (AcoD) example shipped with the toolbox.

```python
from pathlib import Path
import pandas as pd
import biossbd
from biossbd import load_system
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
EXAMPLE = Path("..") / "examples" / "acod" / "acod_system.yaml"
system = load_system(EXAMPLE)
print("bioSSbD-py", biossbd.__version__, "|", system.name)
```

## 1.1 The system and its alternatives
The YAML file declares the functional unit, the life-cycle stages, shared units and three design alternatives (SSbD scenarios).

```python
print(system.functional_unit)
print([s.name for s in system.stages])
print(system.alternative_names)
pd.DataFrame([a.model_dump() for a in system.actors])
```

## 1.2 Hazard register
Every unit contributes hazards from its substances (GHS hazard statements) and from biological or process hazards that GHS does not cover.

```python
from biossbd import hazard_register
reg = hazard_register(system, "Coffee husk + manure")
reg.head(12)
```

## 1.3 Life-cycle safety matrix
Maximum screening severity (1–5) per stage and hazard group. A row of zeros means *either* hazard-free *or* not yet assessed – the matrix makes that question visible.

```python
from biossbd import safety_matrix, receptor_matrix
safety_matrix(system, "Coffee husk + manure")
```

```python
receptor_matrix(system, "Coffee husk + manure")
```

## 1.4 Scoping gaps

```python
from biossbd import scoping_gaps
scoping_gaps(system)
```
